```python
import pandas as pd

# Creating sample employee data
data = {
    "name": ["Aman", "Riya", "Neha", "Karan", "Pooja", "Rahul", "Sneha", "Arjun",
             "Kriti", "Vikas", "Anita", "Rohit"],
    "age": [22, 21, 23, 24, 22, 25, 21, 23, 22, 24, 26, 23],
    "salary": [25000, 27000, 26000, 30000, 28000, 32000, 24000,
               29000, 27500, 31000, 33000, 29500]
}

df = pd.DataFrame(data)

# Save DataFrame to CSV
df.to_csv("employee_data.csv", index=False)

# Load the CSV file
loaded_df = pd.read_csv("employee_data.csv")
loaded_df
```

```python
# Save file with colon separator
loaded_df.to_csv("colon_file.txt", sep=":", index=False)

# Read colon separated file
colon_df = pd.read_table("colon_file.txt", sep=":")
colon_df
```
