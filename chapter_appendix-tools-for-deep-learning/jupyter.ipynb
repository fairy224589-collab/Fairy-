```python
#Name- fairy
#Enrollment no. - 03701032025
#subject - IDS LAB
#B.tech(IT-1)
#lab-17

import numpy as np
import matplotlib.pyplot as plt

#Q1(a). Plot the following functions using Matplotlib, each on a separate graph:
#use x = 1 to 20

# x from 1 to 20
x = np.arange(1, 21)

# 1) 3x - 4
plt.plot(x, 3*x - 4)
plt.title("3x - 4")
plt.grid()
plt.show()

# 2) x^2 + 2x - 15
plt.plot(x, x**2 + 2*x - 15)
plt.title("x^2 + 2x - 15")
plt.grid()
plt.show()

# 3) 5(x-1)(x-2)(x-3)
plt.plot(x, 5*(x-1)*(x-2)*(x-3))
plt.title("5(x-1)(x-2)(x-3)")
plt.grid()
plt.show()

# 4) e^x
plt.plot(x, np.exp(x))
plt.title("e^x")
plt.grid()
plt.show()


#Q1(b)
#use x = 1 to 100
#f(x)=logx

x = np.arange(1, 101)

plt.plot(x, np.log10(x))
plt.title("log(x)")
plt.grid()
plt.show()


#Q2. Plot trigonometric functions on separate subplots

x = np.linspace(-2*np.pi, 2*np.pi, 100)

plt.subplot(2,2,1)
plt.plot(x, np.sin(x))
plt.title("sin(x)")
plt.grid()

plt.subplot(2,2,2)
plt.plot(x, np.cos(x))
plt.title("cos(x)")
plt.grid()

plt.subplot(2,2,3)
plt.plot(x, np.tan(x))
plt.title("tan(x)")
plt.ylim(-10,10)
plt.grid()

plt.subplot(2,2,4)
plt.plot(x, 1/np.tan(x))
plt.title("cot(x)")
plt.ylim(-10,10)
plt.grid()

plt.tight_layout()
plt.show()
```
